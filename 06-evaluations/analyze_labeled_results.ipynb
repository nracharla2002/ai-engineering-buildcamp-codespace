# Count bad evaluation labels

Read `results_labeled.json` and count results labeled `bad`.

```python
import json
from pathlib import Path

results_path = Path('results_labeled.json')
if not results_path.exists():
    raise FileNotFoundError(f'Could not find {results_path.resolve()}. Run this notebook from the 06-evaluations folder, or update results_path.')

with results_path.open(encoding='utf-8') as f:
    results = json.load(f)

bad_results = [row for row in results if row.get('label') == 'bad']
labeled_results = [row for row in results if row.get('label') in {'good', 'bad'}]

print(f"Bad results: {len(bad_results)}")
print(f"Labeled results: {len(labeled_results)} of {len(results)}")
```

```python
# Optional: inspect the questions labeled bad.
for row in bad_results:
    print('-', row.get('question', '(missing question)'))
    if row.get('comments'):
        print('  Comment:', row['comments'])
```
